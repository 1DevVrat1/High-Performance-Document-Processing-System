```python
import fitz  # PyMuPDF for fast PDF processing

def extract_text_from_pdf(pdf_path):
    """Extracts text from a PDF file and returns as a string."""
    try:
        doc = fitz.open(pdf_path)
        text = ""
        for page in doc:
            text += page.get_text("text") + "\n"
        return text
    except Exception as e:
        return f"Error processing {pdf_path}: {e}"

# Set the PDF path in Colab
pdf_path = "/content/DEV_VRAT_RESUME.pdf"

# Extract and display text
pdf_text = extract_text_from_pdf(pdf_path)
print(pdf_text[:1000])  # Print first 1000 characters to preview
```

```python
from groq import Groq
import getpass  # For securely entering API key

# Ask user for Groq API key securely (hides input)
API_KEY = getpass.getpass("Enter your Groq API Key: ")

def query_groq_llm(document_text, user_question):
    """Queries Groq LLM based on the extracted text from the PDF."""
    client = Groq(api_key=API_KEY)

    response = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[
            {"role": "system", "content": "Answer the question based only on the given document."},
            {"role": "user", "content": f"Document Text: {document_text[:3000]}\n\nUser Question: {user_question}"}
        ]
    )

    return response.choices[0].message.content

# Ask user for their question
user_question = input("Enter your question about the PDF: ")

# Ensure `pdf_text` is loaded (You need to extract text from the PDF first)
if 'pdf_text' not in globals():
    print("Error: No PDF text found. Please extract text from your PDF first.")
else:
    response = query_groq_llm(pdf_text, user_question)
    print("\nLLM Response:\n", response)
```
